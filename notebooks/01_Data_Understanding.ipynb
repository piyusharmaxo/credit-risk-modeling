```python
# Importing the required libraries.
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

#Display all columns and wider tables.
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 1000)
```

```python
# Loading the raw application dataset.
PROJECT_ROOT = Path.cwd().parent
DATA_PATH = PROJECT_ROOT / "data" / "raw" / "application_train.csv"

if not DATA_PATH.exists():
    raise FileNotFoundError(f"Dataset not found at: {DATA_PATH}")

df = pd.read_csv(DATA_PATH)

print(f"Number of observations: {df.shape[0]:,}")
print(f"Number of variables: {df.shape[1]:,}")

display(df.head())
display(df.sample(5, random_state=42))
```

```python
# Variable types and initial data-quality assessment.
numeric_cols = df.select_dtypes(include=["int64", "float64"]).columns

categorical_cols = df.select_dtypes(include=["object"]).columns

print("Numerical variables:", len(numeric_cols))
print("Categorical variables:", len(categorical_cols))

unique_summary = pd.DataFrame({"Data Type": df.dtypes,
                               "Unique Values": df.nunique(),
                               "Missing Values": df.isnull().sum(),
                               "Missing %": (df.isnull().mean() * 100).round(2)})

print("\nComplete variable-level data-quality summary:")
display(unique_summary)

duplicate_count = df.duplicated().sum()

print(f"\nDuplicate rows: {duplicate_count:,}")
```

```python
# Identifying the identifier and target variables.
identifier_columns=["SK_ID_CURR"]
target_column= "TARGET"

print("Identifier:", identifier_columns)
print("Target:", target_column)
```

## Identifier and Target Treatment

`SK_ID_CURR` is an application-level identifier rather than an economically meaningful borrower characteristic. It is retained for record-level traceability but excluded from the machine-learning feature matrix.

`TARGET` is the binary credit-risk outcome:

- `0` = no observed default
- `1` = observed default

The target is highly imbalanced, with defaults representing a minority of applications. Consequently, subsequent model development will evaluate both ROC-AUC and precision-recall performance rather than relying on accuracy alone.

```python
# Target variable distribution
target_counts = df["TARGET"].value_counts()
target_percent = df["TARGET"].value_counts(normalize=True) * 100
target_summary = pd.DataFrame({"Count": target_counts, "Percentage": target_percent.round(2)})
target_summary
```
